```python
def lb_to_kg(pounds):
  return pounds * 0.453592

print(f'10 lb is {lb_to_kg(100):.1f} kg')
print(f'200 lb is {lb_to_kg(150):.1f} kg')

my_weight_lb = 165
print(f'{my_weight_lb} lb is {lb_to_kg(my_weight_lb):.1f} kg')
```
