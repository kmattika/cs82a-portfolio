```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

Step One

```python
df = pd.read_csv(r'C:\Users\kamat\OneDrive\Desktop\CS82A\sales_clean.csv')
display(df.head(10))
display(df.shape)
display(df.info())
display(df.describe())
```

Step Two

```python
print("The mean price is", df['price'].mean())
print("The median price is", df['price'].median())
print("The standard deviation of the price is", df['price'].std())
print("The mean quantity is", df['qty'].mean())
print("The median quantity is", df['qty'].median())
print("The standard deviation of the quantity is", df['qty'].std())
```

Step Three

I would report the median price. From looking at a brief summary of the data, Price is skewed by a large max value of 479.29, nearly six standard deviations away from the mean. As that value pulls the mean value nearly twenty points higher than the median value, the mean has been measurably impacted by this skew. Therefore, the median is the more accurate value to report.
