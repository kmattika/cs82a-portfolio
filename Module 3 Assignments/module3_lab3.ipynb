```python
df[df['length'] >= 3809.944]
```

I will use one half standard deviations as the basis for what constitutes similar data.

```python
df[(df['erected'] > 1878) & (df['erected'] < 1916) & (df['purpose'] == 'RR') & (df['lanes'] == 2.0) & (df['material'] == 'STEEL') & (df['span'] == 'LONG')]
```

We can use our heuristic skills here and examine the list of bridges with lengths of one standard deviation above the mean. Our outlier values, E34 and E46, only have a single other bridge of remotely similar length present. E91, which was constructed in 1975, with a length of 3756 units, consisting of six lanes, and uses an arch-type of construction, cannot plausibly support the presumption that the lengths of E34 and E46 are true. The differences in construction material and technique span nearly a century of time, and given that the positive third standard deviation of the mean has a value of 3809.943, E91 itself is just barely statistically plausible. Thus, I would remove rows 31 and 44 (corresponding to E34 and E46 respectively).

Data cleaning notes, in order:

Removed rows 31, 44, as we cannot trust that data.
Fill in null values found in 'length' with the column's median value.

```python
df = df.drop([31, 44])
fill_value = df['length'].median()
df['length'] = df['length'].fillna(fill_value)
```
