```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv(r'C:\Users\kamat\Downloads\ratings.tsv', sep='\t')
display(df.head(10))
display(df.shape)
display(df.info())
display(df.describe())
```

```python
X = df[['food', 'ambience', 'service']]
y = df['rating']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)

print('X_train:', X_train.shape)
print('X_test:', X_test.shape)
print('y_train:', y_train.shape)
print('y_test:', y_test.shape)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print('Coefficients:', model.coef_)
print('Intercept:', model.intercept_)
print('R-squared:', model.score(X, y))
```

```python
predictions = pd.DataFrame({
    'Actual Rating': y_test,
    'Predicted Rating': y_pred
})

display(predictions.round(3))
```

```python
coefficient = pd.DataFrame({
    'Factor': X.columns,
    'Coefficient': model.coef_
})

coefficient = coefficient.sort_values(
    by='Coefficient',
    ascending=False
)

display(coefficient.round(3))
```

From the analysis present, food is the overwhelming factor that influences ratings for a restaurant in our model. This seems to reflect a real-world shift of the business case in the food industry - consumers are consuming prepared food in more modalities than a decade ago, with food delivery apps and food trucks capturing more market engagement during and after the covid-19 pandemic. Thus, we can theorize that service (which can be outsourced to third party delivery apps, or be as simple as waiting for food to be served at a kiosk to-go) and ambience (which is cut out entirely with food delivery and to-go dining options) are overwhelmingly inferior factors to the quality of the food itself.
