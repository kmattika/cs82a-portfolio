```python
value_a = int(input("Please provide the first whole number to multiply: "))
value_b = int(input("Please provide the second whole number to multiply: "))
```

```python
# Runs the 'while' loop for multiplying values without the use of the multiplication operators

def multiplication_while(value_a, value_b):
    answer = 0
    counter = value_b

    while counter != 0:
        answer = answer + value_a
        counter = counter - 1

    return answer
```

```python
# Runs the 'for' loop for multiplying values without the use of the multiplication operators

def multiplication_for(value_a, value_b):
    answer = 0

    for counter in range(value_b):
        answer = answer + value_a

    return answer
```

```python
# Manually checks and displays the result of each loop's results for the user

while_answer = multiplication_while(value_a, value_b)
for_answer = multiplication_for(value_a, value_b)

print("While loop result:", while_answer)
print("For loop result:", for_answer)
print("Real product:", value_a * value_b)
```

```python
# Displays results of checking matching loop values. Did this section as warmup for if / else conditionals,
# plus this is a more user friendly way of satisfying the problem's conditions

if while_answer == for_answer:
    print("The products of both loops match. Success!")
else:
    print("The products of both loops do not match. Huh?")
```
