```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import sqlite3
```

```python
con = sqlite3.connect('world.db')

query = '''
   
    CREATE TABLE IF NOT EXISTS Customers (
    ID INTEGER PRIMARY KEY AUTOINCREMENT,
    Name TEXT NOT NULL DEFAULT '',
    City TEXT NOT NULL DEFAULT '',
    ZIP TEXT NOT NULL DEFAULT ''
);
'''

con.execute(query)
con.commit()
```

```python
query = '''

    INSERT INTO Customers (Name, City, Zip)
    VALUES
        ('Adam Apple', 'New York', '10001'),
        ('Betsy Boot', 'Hialeah', '33018'),
        ('Cade Canderson', 'Seattle', '98122'),
        ('Dave Danderson', 'Houston', '77002'),
        ('Ezekiel Endy', 'Boston', '02108'),
        ('Francis French', 'Montpelier', '05601'),
        ('Grogina Granite', 'Billings', '59101'),
        ('Hugh Horse', 'Billings', '59102');

'''

con.execute(query)
con.commit()
```
