```python
import pandas as pd
import sqlite3
import seaborn as sns
import matplotlib.pyplot as plt
con = sqlite3.connect('world.db')
```

```python
pd.read_sql(
    '''
    SELECT *
    FROM Customers
    WHERE city LIKE '%i%'
    ''',
    con
)
```

```python
pd.read_sql(
    '''
    SELECT *
    FROM Customers
    WHERE City = 'Billings'
    ORDER BY Name DESC
    ''',
    con
)
```

```python
df = pd.read_sql(
    '''
    SELECT *
    FROM Customers
    WHERE zip > 10000
    ORDER BY ZIP ASC
    ''',
    con
)
```

```python
sns.scatterplot(
    data=df,
    x='City',
    y='ZIP'
)

plt.title('No relationship between city name and zip code, despite appearance of trendline')
plt.xlabel('City')
plt.ylabel('Zip')
plt.show()
```
