```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind
```

```python
df = pd.read_csv(r'C:\Users\kamat\OneDrive\Desktop\CS82A\custdata.tsv', sep="\t")
display(df.head(10))
display(df.shape)
display(df.info())
display(df.describe())
```

```python
df.dropna(subset=["housing.type"], inplace=True)
display(df.head(10))
display(df.shape)
```

```python
counts = df['housing.type'].value_counts().reset_index()
counts.columns = ['housing.type', 'count']

sns.barplot(
    data=counts,
    x='housing.type',
    y='count',
    order=counts['housing.type'],
    color='salmon'
)

plt.title('Most of our sample population regularly pays for their dwelling')
plt.xlabel('Housing Type')
plt.ylabel('Sample Population')
plt.xticks(rotation=30)
plt.show()
```
