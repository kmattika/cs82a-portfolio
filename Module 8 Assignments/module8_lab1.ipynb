```python
# Import the libraries needed for the analysis.
# pandas handles tables and calculations.
# seaborn creates the chart.
# matplotlib controls the chart's size, title, and axis labels.
# display presents DataFrames as readable tables in the notebook.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from IPython.display import display


# Read the CSV into a DataFrame named df.
# The filename is relative to the notebook's current working folder.
# If this raises FileNotFoundError, check that penguins.csv is uploaded
# and that its spelling and capitalization match the filename below.
df = pd.read_csv("penguins.csv")

# Stop with an explanatory message if the CSV contains no data rows.
# This also prevents calculating percentages with a zero denominator.
if df.empty:
    raise ValueError("penguins.csv contains no data rows.")

print(f"Dataset size: {len(df)} rows and {len(df.columns)} columns")


# ----------------------------------------------------------
# SUMMARIZE MISSING VALUES FOR EVERY COLUMN
# ----------------------------------------------------------

# isna() creates a table with the same shape as df.
# Each missing cell becomes True; each non-missing cell becomes False.
missing_mask = df.isna()

# Summing down each column counts its True values because
# True behaves like 1 and False behaves like 0 in this calculation.
missing_counts = missing_mask.sum()

# Build a summary with one row per original column.
# Each percentage uses the total number of dataset rows as its denominator.
# Round percentages for readability, without changing the original data.
missing_summary = pd.DataFrame({
    "non_missing_count": len(df) - missing_counts,
    "missing_count": missing_counts,
    "missing_percent": (missing_counts / len(df) * 100).round(2)
})

print("\nMissing data by column:")
display(missing_summary)


# Count rows containing at least one missing cell.
# any(axis=1) checks across the columns within each row.
# A row with several missing cells is counted only once here.
rows_with_missing = missing_mask.any(axis=1).sum()

print(
    f"Rows with at least one missing value: "
    f"{rows_with_missing} of {len(df)} "
    f"({rows_with_missing / len(df) * 100:.2f}%)"
)

# Do not drop every row with any missing data.
# A penguin with missing sex can still contribute to a body-mass
# comparison when its body_mass_g value is available.
```

```python
# ----------------------------------------------------------
# COMPARE COLUMNS 2–5: NUMERICAL MEASUREMENTS
# ----------------------------------------------------------

# Use the actual column names rather than their positions.
# This keeps the code correct if the CSV's columns are reordered.
measurement_cols = [
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]

# groupby("species") separates the rows into species groups.
# Selecting measurement_cols limits the calculations to the four
# requested body measurements.
#
# agg() applies several summary calculations to every measurement:
# count  = number of non-missing values used for that measurement
# mean   = arithmetic average
# median = middle value after sorting the available observations
# std    = sample standard deviation, measuring spread around the mean
# min    = smallest available value
# max    = largest available value
#
# These calculations exclude missing values in each measurement
# individually. A missing bill length does not remove an otherwise
# available body mass from the body-mass calculation.
measurement_summary = (
    df.groupby("species")[measurement_cols]
      .agg(["count", "mean", "median", "std", "min", "max"])
)

# The result contains statistics for all four measurements.
# Display one measurement at a time so the tables remain narrow
# and easy to read in Colab. Each table has one row per species.
for column in measurement_cols:
    print(f"\n{column} by species:")
    display(measurement_summary[column].round(2))


# ----------------------------------------------------------
# COMPARE COLUMN 6: SEX
# ----------------------------------------------------------

# Sex is categorical, so compare frequencies rather than averages.
#
# Create a separate Series that labels missing sex as "Missing".
# This keeps missing records visible in the frequency table.
# It does not change df["sex"] or guess the penguins' actual sex.
sex_for_summary = df["sex"].fillna("Missing")

# crosstab() counts records for each combination of species and sex.
# Rows represent species; columns represent the sex labels in the CSV,
# plus "Missing" when missing sex values are present.
sex_counts = pd.crosstab(
    index=df["species"],
    columns=sex_for_summary
)

# sum(axis=1) adds across the sex categories for each species.
# div(..., axis=0) divides each row by its corresponding row total.
# Multiplying by 100 converts proportions into percentages.
#
# The denominator includes records labeled "Missing", so these are
# percentages of ALL records for each species, not just known-sex records.
sex_percentages = (
    sex_counts.div(sex_counts.sum(axis=1), axis=0) * 100
).round(2)

print("\nSex counts by species, including missing sex:")
display(sex_counts)

print("\nSex percentages within each species:")
display(sex_percentages)


# ----------------------------------------------------------
# COMPARE COLUMN 7: YEAR
# ----------------------------------------------------------

# Although year is stored as an integer, it identifies an observation
# period rather than a body measurement.
# Compare the number of observations in each year for each species.
# Your supplied schema indicates that year has no missing values.
year_counts = pd.crosstab(
    index=df["species"],
    columns=df["year"]
)

# Calculate the percentage of each species' records observed in each year.
# These percentages help compare sampling patterns even when the species
# have different total numbers of observations.
year_percentages = (
    year_counts.div(year_counts.sum(axis=1), axis=0) * 100
).round(2)

print("\nObservation counts by species and year:")
display(year_counts)

print("\nYear percentages within each species:")
display(year_percentages)

# Each row in either percentage table should sum to approximately 100%.
# Small differences from 100% can result from rounding.
```

```python
# ----------------------------------------------------------
# CREATE ONE CHART: BODY MASS BY SPECIES
# ----------------------------------------------------------

# Keep rows with the two values required for this chart.
# Missing values in unrelated columns, such as sex or bill length,
# do not cause a record to be excluded from this plot.
#
# dropna() returns a separate DataFrame here; df is left unchanged.
plot_df = df.dropna(subset=["species", "body_mass_g"])

# Stop with a useful message if there are no usable plotting records.
if plot_df.empty:
    raise ValueError("No rows have both species and body_mass_g available.")

# Alphabetize the species labels for a consistent x-axis order.
species_order = sorted(plot_df["species"].unique())

# Create one figure and one plotting area.
# figsize specifies the figure's width and height in inches.
fig, ax = plt.subplots(figsize=(9, 6))

# Draw a separate body-mass box for each species.
#
# The line inside each box is the median.
# The box extends from the 25th to the 75th percentile.
# Its height is the interquartile range (IQR).
#
# whis=1.5 extends each whisker to the most extreme observed value
# still within 1.5 * IQR of the corresponding edge of the box.
# Values beyond the whiskers are shown as individual points.
# These points are not automatically errors and are not deleted.
sns.boxplot(
    data=plot_df,
    x="species",
    y="body_mass_g",
    order=species_order,
    whis=1.5,
    ax=ax
)

# Add a descriptive title and label both axes.
# Include the measurement unit so the y-axis is unambiguous.
ax.set_title("Distribution of Penguin Body Mass by Species")
ax.set_xlabel("Penguin species")
ax.set_ylabel("Body mass (grams)")

# Adjust spacing so the labels fit, then display the single chart.
fig.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# This code uses the DataFrame named df from your previous analysis.
# If you restarted your Colab session, uncomment the following line:
# df = pd.read_csv("penguins.csv")


# ----------------------------------------------------------
# STEP 1: SELECT ADELIE PENGUINS AND THE REQUIRED COLUMNS
# ----------------------------------------------------------

# Store the four measurement column names in a list.
# We will use this list both to select data and to order the panels.
measurement_cols = [
    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
]

# df["species"] == "Adelie" identifies rows for the requested species.
#
# .loc[row_condition, columns] selects those rows and keeps only
# island plus the four measurements.
#
# .copy() creates a separate DataFrame so the original df is unchanged.
adelie_df = df.loc[
    df["species"] == "Adelie",
    ["island"] + measurement_cols
].copy()


# ----------------------------------------------------------
# STEP 2: RESHAPE THE DATA FOR A MULTI-PANEL GRAPH
# ----------------------------------------------------------

# The original data is in "wide" format:
# one row per penguin, with a separate column for each measurement.
#
# melt() changes it to "long" format:
# island       measurement         value
# <island>     bill_length_mm       <bill length>
# <island>     bill_depth_mm        <bill depth>
# <island>     flipper_length_mm    <flipper length>
# <island>     body_mass_g          <body mass>
#
# Each penguin contributes up to four usable measurement rows.
# These rows represent different measurements, not different penguins.
#
# id_vars preserves the island associated with each measurement.
# value_vars identifies the columns to turn into measurement rows.
# var_name names the new column containing measurement names.
# value_name names the new column containing numerical values.
adelie_long = adelie_df.melt(
    id_vars="island",
    value_vars=measurement_cols,
    var_name="measurement",
    value_name="value"
)

# Remove missing measurements AFTER reshaping.
# This means a penguin missing bill length can still contribute its
# available bill depth, flipper length, and body mass to other panels.
#
# Records also need an island so they can be assigned to a group.
# Missing sex or year does not affect inclusion in this graph.
adelie_long = adelie_long.dropna(subset=["island", "value"])

# Stop with an explanatory message if no usable data remains.
if adelie_long.empty:
    raise ValueError(
        "No usable Adelie measurements were found. "
        "Check the species labels and measurement columns in df."
    )


# ----------------------------------------------------------
# STEP 3: CREATE THE FOUR-PANEL BOXPLOT
# ----------------------------------------------------------

# Keep islands in the same order in every panel.
island_order = ["Biscoe", "Dream", "Torgersen"]

# Create readable labels that include measurement units.
measurement_labels = {
    "bill_length_mm": "Bill length (mm)",
    "bill_depth_mm": "Bill depth (mm)",
    "flipper_length_mm": "Flipper length (mm)",
    "body_mass_g": "Body mass (g)"
}

# x="island" compares the three islands within each panel.
# y="value" supplies the numerical measurements.
# col="measurement" creates one panel per measurement.
# col_order preserves the order in measurement_cols.
# col_wrap=2 arranges the four panels into two columns and two rows.
# kind="box" displays each island group's distribution.
#
# IMPORTANT: sharey=False gives each measurement its own y-axis scale.
# Body mass uses grams, whereas the other measurements use millimeters.
# Even the millimeter measurements have different numerical ranges.
g = sns.catplot(
    data=adelie_long,
    x="island",
    y="value",
    col="measurement",
    col_order=measurement_cols,
    col_wrap=2,
    order=island_order,
    kind="box",
    sharey=False,
    height=4,
    aspect=1.2
)


# ----------------------------------------------------------
# STEP 4: LABEL AND DISPLAY THE FIGURE
# ----------------------------------------------------------

# axes_dict links each measurement name to its plotting panel.
# Loop through the panels to give each one a readable title and
# a y-axis label with the correct measurement unit.
for measurement, ax in g.axes_dict.items():
    ax.set_title(measurement_labels[measurement])
    ax.set_xlabel("Island")
    ax.set_ylabel(measurement_labels[measurement])

    # Show the island names on every panel, including the top row.
    ax.tick_params(axis="x", labelbottom=True)

# Add a title covering the entire figure.
g.figure.suptitle(
    "Adélie Penguin Measurements Across the Three Islands",
    fontsize=15
)

# Adjust spacing and reserve room above the panels for the main title.
g.figure.tight_layout(rect=[0, 0, 1, 0.94])

plt.show()
```

```python

bill_df = df.loc[
    df['bill_depth_mm'] > 0,
    ['species', 'bill_length_mm', 'bill_depth_mm']
].dropna().copy()

bill_df['bill_ratio'] = (
    bill_df['bill_length_mm'] / bill_df['bill_depth_mm']
)

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=bill_df,
    x='species',
    y='bill_ratio'
)

plt.title('Penguin Bill Proportions by Species')
plt.xlabel('Penguin species')
plt.ylabel('Bill length / bill depth (ratio)')
plt.tight_layout()
plt.show()
```

Here, we can see that Adelie penguins have more compact bill proportions compared to Chinstrap and Gentoo penguins.
